```python
import platform
import ipywidgets as widgets
from ipywidgets import VBox
from pathlib import Path
from fastai.vision.all import *


learn_inf = load_learner("export.pkl")

btn_upload = widgets.FileUpload()
out_pl = widgets.Output()
out_pl.clear_output()
lbl_pred = widgets.Label()
btn_run = widgets.Button(description="Classify")

def on_click_classify(change):
    assert len(btn_upload.value) > 0, "No images uploaded, gimme a new one"
    upload = btn_upload.value[-1]
    raw_bytes = upload['content'].tobytes()
    img = PILImage.create(raw_bytes)
    out_pl.clear_output()
    with out_pl: display(img.to_thumb(128,128))
    pred,pred_idx,probs = learn_inf.predict(img)
    lbl_pred.value = f"Prediction: {pred}, Probability: {probs[pred_idx]:.04f}"
btn_run.on_click(on_click_classify)

VBox([widgets.Label("Gimme da bear or da goat"), btn_upload, btn_run, out_pl, lbl_pred])
```
